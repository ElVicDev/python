## Crea una función que devuelva el nombre de una persona formateado como un saludo

```python
def greet(name):
    """ A greeting message including a person's name """

    # Combina la palabra "Hola" con el nombre de la persona y un signo de exclamación para crear el mensaje de saludo.
    name = "Hello " + str(name) + "!"

    # Devuelve el mensaje de saludo para que pueda usarse en otra parte del código.
    return name
```

### Llama a la función de saludo con el nombre "Alice" y almacena el mensaje de saludo devuelto en la variable `result`

```python
# Llamar a la función de saludo, almacenar el resultado en result
result = greet("Alice")

# Mostrar result
print(result)
```
