# Flujo de control en Python: El conductor de su código.
En Programación informática, el flujo de control dicta qué líneas de código se ejecutan, en qué orden y bajo qué condiciones.

## Sentencias condicionales y bucles: La base del flujo del programa
Dos de las herramientas fundamentales del flujo de control son las **sentencias condicionales** y los **bucles**. Se trata de construcciones de programación esenciales que dictan el modo en que el código toma decisiones y repite tareas.
Las **sentencias condicionales** permiten que el código se haga preguntas como "¿Está la temperatura por encima del punto de congelación?" o "¿Ha introducido el usuario la contraseña correcta?" y, a continuación, elija diferentes caminos en función de las respuestas.
Los **bucles**, por otro lado, permiten que tu código repita un conjunto de instrucciones varias veces, lo que es increíblemente útil para tareas como el procesamiento de grandes cantidades de datos o la realización de acciones repetitivas.
*Sin flujo de control, tus programas Python no podrían hacer mucho más que ejecutar una simple secuencia de instrucciones.*

## Por qué es importante el flujo de control
Imagina que estás creando un sistema de registro de usuarios para un sitio web restringido. Sin flujo de control, recopilaría y almacenaría datos sin pensar, lo que podría dar lugar a cuentas duplicadas o problemas de seguridad.
Con el flujo de control, podemos validar la entrada (por ejemplo, para garantizar contraseñas seguras), comprobar si hay duplicados y proporcionar información al usuario si hay algún problema.

## Sentencias condicionales: Tomar decisiones

Imagine las sentencias condicionales como la lógica "si-entonces-si" de la programación. Permiten al código evaluar una situación y seleccionar una acción en función de si una condición es verdadera o falsa.
Por ejemplo, una aplicación meteorológica podría utilizar condicionales para aconsejar "¡ponte una chaqueta!" si la temperatura desciende por debajo de un cierto punto, o sugerir un atuendo más ligero en caso contrario.

*En Python, la piedra angular de las sentencias condicionales es la sentencia if.*

```python
temperature = 22  # Example temperature

if temperature < 20:
    print("It might be cold! You might need a jacket.")
else:
    print("It's warm! Enjoy the sunshine.")
```

Este código demuestra cómo utilizar las sentencias **if** y **else** para crear una sencilla estructura de toma de decisiones basada en diferentes rangos de temperatura, proporcionando recomendaciones apropiadas para el tiempo.

Puede refinar la toma de decisiones de su código incorporando **elif** (abreviatura de "*else if*") para evaluar varias condiciones secuencialmente. Esto le permite proporcionar mensajes diferentes para temperaturas frías, templadas y cálidas.

```python
temperature = 15  # Let's assume the temperature is 15 degrees Celsius

if temperature < 10:
    print("Wear a jacket! It's cold out there.")
elif 10 <= temperature < 20:
    print("A light sweater should be fine. It's a bit chilly.")
elif 20 <= temperature < 30:
    print("Enjoy the pleasant weather! No need for extra layers.")
else:
    print("It's hot! Stay hydrated and wear sunscreen.")
```

Este código demuestra cómo **elif** le permite encadenar varias condiciones, proporcionando respuestas más matizadas basadas en el valor de la temperatura.

## Bucles: Automatizar la repetición

Los bucles reflejan las tareas repetitivas que nos encontramos en la vida, como enviar correos electrónicos a compañeros de clase o calcular promedios. Permiten ejecutar un bloque de código varias veces, ya sea durante un número determinado de repeticiones o hasta que se cumpla una condición específica.

Python ofrece dos tipos principales de bucles: **for** y **while**.

*Los bucles for son perfectos cuando se conoce de antemano el número de repeticiones deseado.*

```python
# Ejemplo de código: bucle for (iteración definida)
# Example: Print numbers from 1 to 5
for i in range(1, 6):
    print(i)
```

```python
# Ejemplo de código: bucle while (iteración indefinida)
# Example: Keep asking for input until the user enters "quit"
user_input = ""
while user_input != "quit":
    user_input = input("Enter something (or 'quit' to exit): ")
    print("You entered:", user_input)
```
