### Crea una función con una función anidada dentro

```python
def outer_function(x):
    """
    Función externa que toma un entero x, le suma 5 y luego 
    llama a una función interna.

    Argumentos:
    x ​​(int): El valor de entrada que se utilizará en los cálculos.
"""
    y = x + 5  # Suma 5 a x y almacénalo en y

    def inner_function():
        """
        Inner function que multiplica y por 2 e imprime el resultado.
        """
        z = y * 2  # Multiplica y por 2 y almacénalo en z
        print("Inside inner_function, z =", z)  # Imprima el valor de z

    inner_function()  # Call the inner_function
    print("Inside outer_function, y =", y)  # Imprima el valor de y después llamar a inner_function
```

### Llamar a la función externa

```python
# Llama a outer_function con un argumento de 10
outer_function(10)
```
