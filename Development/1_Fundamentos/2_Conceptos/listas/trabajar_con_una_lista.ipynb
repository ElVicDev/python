## Escenario: Nunca más olvides la leche: Tu lista de la compra con Python

Imagina que vas al supermercado, pero siempre olvidas los productos que necesitas. ¡Es frustrante y te obliga a ir varias veces! ¿No sería genial tener una herramienta digital para llevar un registro de tu lista de la compra?

### Por qué las listas son perfectas para las listas de compras:

* **Colección Ordenada:** Las listas mantienen el orden en que agregas artículos, lo cual puede ser útil para priorizar tus compras.
* **Mutable:** Puedes agregar, eliminar o modificar artículos fácilmente según tus necesidades.
* **Flexible:** Las listas pueden almacenar artículos de diferentes tipos (por ejemplo, cadenas para nombres de artículos, números para cantidades).

## Resumen de la actividad:

En esta actividad, aprenderás a:
* Crear y gestionar una lista de la compra dinámica con listas de Python.
* Cargar artículos de un archivo CSV en tu lista de la compra.
* Añadir artículos a la lista con el método `append()`.
* Eliminar artículos de la lista con el método `remove()`.

Para esta actividad, tendrás acceso a un archivo CSV llamado `grocery_list.csv`, ya creado, que contiene algunos artículos iniciales para nuestra lista de la compra.

Comienza ejecutando la celda a continuación, que importa la biblioteca de Pandas, carga el archivo CSV llamado `grocery_list.csv` e imprime el resultado.

```python
# Importa una potente herramienta llamada "pandas" que usarás 
# para trabajar y organizar datos fácilmente
import pandas as pd
import os 

file_name = 'grocery_list.csv'

if os.path.exists(file_name):
    # Cargar la lista de compras desde el archivo CSV
    grocery_list_df = pd.read_csv(file_name)

    # Extrae los elementos del DataFrame y almacénalos en una lista
    grocery_list = grocery_list_df['item'].tolist()

    # Imprima la lista de compras e inspeccione el resultado
    print(grocery_list)
else:
    print(f"Error: The file '{file_name}' was not found in the current directory.")
    print("Please ensure you have followed the instructions on the Coursera site to extract the files from the zip archive.")
```

Verás que tu lista de la compra ya tiene algunos artículos, ¡pero quizás necesites algunos más!

Podemos simular el proceso de recopilar la información del usuario para añadir artículos a tu lista de la compra. Imagina que quieres introducir los siguientes artículos: "Kiwis" y "Frambuesas".

Añade "Kiwis" y "Frambuesas" a la lista `items_to_add` y luego ejecuta la celda.

```python
# Añade "Kiwis" y "Frambuesas" a la lista.
items_to_add = ["Kiwis", "Raspberries"]

# Añadir artículos a la lista de compras
for item in items_to_add:
    grocery_list.append(item) 

# Imprima la lista de compras para ver los artículos agregados
print(grocery_list) 
```

Observa que "Kiwis" y "Frambuesas" se han añadido al final de tu lista de la compra.

¿Y si quieres añadir un artículo diferente?

Abre la celda de abajo, donde podrás introducir el artículo que desees. Introduce tu artículo favorito en el cuadro de entrada y pulsa "Intro".

```python
# 1. Obtener la entrada del usuario
new_item = input("Enter an item to add to your grocery list: ")  # Solicitar un artículo al usuario y almacenar su entrada

# 2. Añade el artículo a la lista
grocery_list.append(new_item)  # Añade el nuevo artículo al final de la lista de compras.

# 3. Imprima la lista actualizada
print("Updated list:", grocery_list)  # Mostrar al usuario la lista de compras actualizada
```

Observa que la entrada anterior se ha añadido a tu lista de la compra.

Intenta añadir algunos artículos a la `grocery_list` con el método `append()`.

Añade "Canela" y "Pimentón" al método `append()` (abajo) y ejecuta la celda.

```python
# Añade "Canela" y "Pimentón" a la lista de la compra.
grocery_list.append("Cinnamon") 
grocery_list.append("Paprika") 

# Imprima la lista de compras actualizada
print("\nYour Updated Grocery List:")
print(grocery_list)
```

Observa en el resultado anterior que has añadido varios artículos a tu lista de la compra de diferentes maneras.

Pero te das cuenta de que ya tienes algo en casa. ¿Cómo puedes eliminar artículos de tu `grocery_list`?

Por ejemplo, imagina que ya no necesitas "Huevos" ni "Manzanas". Puedes eliminar estos artículos de la `grocery_list` con el método `remove()`.

Añade "Huevos" y "Manzanas" al método `remove()` que aparece a continuación y ejecuta la celda.


```python
# Elimine "huevos" y "manzanas" de la lista de compras.
grocery_list.remove("Eggs")
grocery_list.remove("Apples") 

# Imprima la lista de compras actualizada
print("\nYour Updated Grocery List (after removing items):")
print(grocery_list)
```

Observa que en la lista de la compra actualizada, se han eliminado los elementos "Huevos" y "Manzanas".
