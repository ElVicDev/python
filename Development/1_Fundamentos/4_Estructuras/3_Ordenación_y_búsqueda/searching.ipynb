¡El código de la celda siguiente hace precisamente eso! Antes de ejecutarlo, vamos a analizarlo:

* **Funciones auxiliares:** En programación, una función auxiliar es simplemente una función que ayuda a otra función a completar su tarea. Piensa en ellas como pequeños ayudantes de biblioteca. Realizan tareas específicas para facilitar la tarea principal (ordenar el catálogo).

```python
# Helper function 1: 
def get_title(book): 
  """Helper function to extract the title from a book dictionary."""
  return book['title']

# Helper function 2: 
def sort_catalog_by_title(catalog): 
  """Sorts the book catalog alphabetically by title."""
  catalog.sort(key=get_title)
    
# Sort the catalog
sort_catalog_by_title(book_catalog)

# Display the sorted catalog
for book in book_catalog:
  print(f"Title: {book['title']}, Author: {book['author']}, Publication Year: {book['publication_year']}")
```

Pero ¿qué pasa si quieres encontrar un libro sin revisar toda la lista? Ahí es donde los algoritmos de búsqueda resultan útiles. Te ayudan a encontrar elementos específicos dentro de un conjunto de datos, igual que un bibliotecario te ayuda a encontrar un libro en el estante.

En la siguiente celda, exploraremos un algoritmo llamado **búsqueda lineal**. Es como revisar sistemáticamente cada libro de un estante hasta encontrar el que buscamos. Pero, ¿cómo se traduce esto a código Python?

Analicemos el código de la siguiente celda antes de ejecutarlo:

La función `search_books` que se muestra a continuación realiza lo siguiente:

1. **Recibe la entrada:** Necesita dos elementos para funcionar:

* El `catalog`: Esta es nuestra lista ordenada de diccionarios de libros.

* La `query`: Esto es lo que buscamos; puede ser el título de un libro o el nombre de un autor.

2. **Crea una lista para los resultados:** Crea una lista vacía llamada `results` para almacenar los libros que coincidan con nuestra búsqueda.

3. **Revisa cada libro:** Recorre cada libro del catálogo:

* Convierte la consulta, el título y el autor del libro a minúsculas. Esto hace que la búsqueda no distinga entre mayúsculas y minúsculas (por lo que "el gran gatsby" y "El Gran Gatsby" se consideran lo mismo).

* Comprueba si la consulta se encuentra en el título o en el autor del libro.

* Si hay una coincidencia, añade ese libro a la lista de resultados.

4. **Devuelve los resultados:** Finalmente, nos devuelve la lista de resultados, que contiene todos los libros que coinciden con nuestra búsqueda (o una lista vacía si no se encuentra ninguno).

**En resumen, es como revisar cada libro en la estantería, comprobar si es el que buscamos y apartarlo si lo es.**

**¿Quieres verlo en acción? ¡Ejecuta la celda de abajo!**

```python
# Helper function: Linear Search
def search_books(catalog, query):
  """Searches for books by title or author using linear search."""
  results = []
  for book in catalog:
    if query.lower() in book['title'].lower() or query.lower() in book['author'].lower():
      results.append(book)
  return results
```

Por suerte, existe una forma más rápida cuando nuestro catálogo está ordenado: la **búsqueda binaria**. Es como un juego de adivinanzas donde reduces a la mitad las posibilidades hasta encontrar lo que buscas. ¿Pero cómo se traduce esto a código Python?

Analicemos el código en la siguiente celda antes de ejecutarlo:

Aquí está lo que hace la función `binary_search_books`:

1. **Recibe la entrada:** Al igual que la búsqueda lineal, necesita el `catálogo` y la `consulta` (el título del libro que buscamos).

2. **Establece los límites:** Comienza marcando el principio (`low`) y el final (`high`) del catálogo.

3. **El juego de adivinanzas:**

* Calcula el punto medio (`mid`) entre `low` y `high`.

* Comprueba si el libro en el punto medio es el que buscamos.

* Si no lo es, compara el título del libro en el punto medio con nuestra consulta:

* Si el título en el punto medio es menor que nuestra consulta, sabemos que el libro que buscamos debe estar en la mitad superior del catálogo, así que ajustamos `low` a `mid + 1`.

* Si el título en el punto medio es mayor que nuestra consulta, sabemos que el libro que buscamos debe estar en la mitad inferior, así que ajustamos `high` a `mid - 1`.

* Repite este proceso, reduciendo el área de búsqueda a la mitad cada vez, hasta que encuentra el libro o determina que no está.

4. **Devuelve el resultado:** Si encuentra el libro, devuelve el diccionario completo. De lo contrario, devuelve `None` para indicar que no se encontró el libro.

**Imagínalo así:** Estás buscando una palabra en un diccionario. En lugar de empezar desde el principio, lo abres por la mitad. Si la palabra que buscas aparece *después* de las palabras de esa página, pasas a la segunda mitad del diccionario. Si aparece *antes*, pasas a la primera mitad. Repites este proceso hasta encontrar la página exacta.

**¿Quieres verlo en acción? ¡Ejecuta la celda de abajo!**

```python
def binary_search_books(catalog, query):
  """Searches for books by title using binary search (assuming sorted catalog)."""
  low = 0
  high = len(catalog) - 1

  while low <= high:
    mid = (low + high) // 2
    if catalog[mid]['title'].lower() == query.lower():
      return catalog[mid]
    elif catalog[mid]['title'].lower() < query.lower():
      low = mid + 1
    else:
      high = mid - 1

  return None  # Book not found
```

En las celdas anteriores, usaste la búsqueda lineal para encontrar "El Gran Gatsby", pero ¿puedes encontrarlo aún *más rápido*?

Recuerda que la búsqueda binaria es extremadamente rápida cuando los datos están ordenados. Funciona dividiendo repetidamente el área de búsqueda por la mitad, como un juego de adivinanzas de alta tecnología.

```python
# This is like using a stopwatch to see how long it takes to find the books. 
# We start the timer, do the search, then stop the timer and calculate how much time passed.
import time

# Loads the big_book_catalog which has which has ~271,380 rows!
big_book_catalog_df = pd.read_csv('big_book_catalog.csv', low_memory=False)

# Convert 'title' and 'author' to strings, handle NaN
big_book_catalog_df['title'] = big_book_catalog_df['title'].fillna('').astype(str)
big_book_catalog_df['author'] = big_book_catalog_df['author'].fillna('').astype(str)

sorted_df = big_book_catalog_df.sort_values(by=['title'])
big_book_catalog = sorted_df.to_dict(orient='records')

# Search for "The Great Gatsby" using Binary Search
query = "The Great Gatsby"  # Example query

start_time = time.time()

search_results = binary_search_books(big_book_catalog, query)

elapsed_time_binary = time.time() - start_time

print(f"\nBinary search took {elapsed_time_binary:.5f} seconds.")

if search_results:
  print("\nBinary Search Result for 'The Great Gatsby':")
  print(f"Title: {search_results['title']}, Author: {search_results['author']}, Publication Year: {search_results['publication_year']}")
else:
  print("Binary Search: Book not found")
```
