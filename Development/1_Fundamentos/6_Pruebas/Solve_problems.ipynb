### Actividad 1: Toma de decisiones con declaraciones condicionales

#### Tarea 1: Definir una función de descuento
Debe escribir una función que ayude al equipo de marketing a decidir qué clientes deben recibir un descuento.

Defina una función llamada ```send_discount``` que acepte dos argumentos:
- ```books_purchased```: La cantidad de libros que ha comprado un cliente.

- ```discount_threshold```: El número mínimo de libros que un cliente debe comprar para recibir un descuento.

### Actividad 2: Uso de bucles para tareas repetitivas
#### Introducción
Eres gerente de producto en una startup tecnológica. Los clientes han estado calificando la última aplicación de la compañía. El equipo de marketing quiere categorizar estos comentarios en tres categorías: "Low", "Medium" y "High". Utilizarás las técnicas de iteración de Python para procesar las calificaciones de los clientes de forma eficiente.
#### Tarea 1: Categorizar las calificaciones de los clientes
Define una función llamada `categorize_ratings` que tome como entrada una lista de calificaciones de los clientes. Cada calificación es un número entero entre 1 y 10.

La función categorizará las calificaciones como:
- **Low** (1-4)
- **Medium** (5-7)
- **High** (8-10)

La salida debe imprimir tres sentencias, una para cada categoría, en el siguiente orden (Low, then Medium, then High):
<br>`Low: {number_of_low_ratings}`<br>
`Medium: {number_of_medium_ratings}`<br>
`High: {number_of_high_ratings}`<br>

#### Ejemplo:
```python
# Hay dos calificaciones en el rango 1-4, dos calificaciones en el rango 5-7 y dos calificaciones en el rango 8-10
categorize_ratings([1, 3, 5, 7, 8, 9])
```
**Output:**
<br>`Low: 2`<br>
`Medium: 2`<br>
`High: 2`<br>
#### Code Template:

````python
def categorize_ratings(rating_list):
    """ Clasifica una lista de calificaciones (1-10) en tres categorías:
        Low, Medium y High.
        Args:
        rating_list (list): Una lista de calificaciones de clientes (enteros)."""
# Inicializar contadores
    low_count = 0    # 1-4
    medium_count = 0 # 5-7
    high_count = 0   # 8-10

# Iterar sobre cada calificación en la lista
    for rating in rating_list:
    # Lógica de categorización:
        if 1 <= rating <= 4:
            low_count += 1
        elif 5 <= rating <= 7:
            medium_count += 1
        elif 8 <= rating <= 10:
            high_count += 1
    # Cualquier otro valor (como 0 o 11+) es ignorado/no contado.

# Imprimir los resultados
    print(f"Low: {low_count}")
    print(f"Medium: {medium_count}")
    print(f"High: {high_count}")
````

#### Consejos:
<details>
<summary>Consejo 1</summary>

- Usa un bucle `for` para iterar por la `rating_list`.
</details>

<details>
<summary>Consejo 2</summary>

- Mantén un contador para cada categoría: baja, media y alta.
</details>

<details>
<summary>Consejo 3</summary>

- Usa `if`, `elif` y `else` para determinar a qué categoría pertenece cada calificación.
</details>

#### Prueba tu función:

#### Paso 4: Ordenar las puntuaciones con una función personalizada
Defina una función llamada `bubble_sort` que ordene la lista de puntuaciones en orden ascendente.

````python
### YOUR CODE HERE ###
# Lista de puntuaciones obtenida en el Paso 3
scores = [87, 90, 75, 100, 100, 70]

def bubble_sort(arr):
    """ Ordena una lista de números en orden ascendente utilizando el algoritmo de Bubble Sort.
        Modifica la lista de entrada (arr) in situ.

        Args:
        arr (list): La lista de puntuaciones a ordenar. """
    
    n = len(arr)

    # Bucle exterior: Itera a través de todos los elementos de la lista.
    # El rango (n-1) es suficiente porque el último elemento ya estará
    # en su lugar después de la primera pasada (pass).
    for i in range(n - 1):
    # Bandera para optimización: si no hubo intercambios en una pasada, 
    # la lista ya está ordenada.
        swapped = False

    # Bucle interior: Compara elementos adyacentes.
    # El rango (n - 1 - i) asegura que no se comprueben los elementos
    # que ya están en su posición final (los 'i' últimos elementos).
    for j in range(0, n - 1 - i):

    # Si el elemento actual es mayor que el siguiente, se intercambian
        if arr[j] > arr[j + 1]:
        # Intercambio (swap) usando la asignación de tuplas de Python
            arr[j], arr[j + 1] = arr[j + 1], arr[j]
        swapped = True

        # Optimización: Si no hubo intercambios en la pasada exterior, 
        # la lista ya está ordenada.
        if not swapped:
            break

# Llamar a la función para ordenar la lista 'scores'
bubble_sort(scores)

print("Lista de Puntuaciones Ordenada (Ascendente):")
print(scores) 
# Resultado esperado: [70, 75, 87, 90, 100, 100]
````

#### Paso 5: Asignar las puntuaciones ordenadas a `sorted_scores`
Llamar a la función `bubble_sort` definida anteriormente y asignar el valor de retorno a `sorted_scores`.

````python
### YOUR CODE HERE ###
# Recreamos la lista original de puntuaciones (antes de ordenar)
scores = [87, 90, 75, 100, 100, 70]

# --- Función bubble_sort del Paso 4 (incluida para contexto) ---
def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        swapped = False
    for j in range(0, n - 1 - i):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]
        swapped = True
        if not swapped:
            break
# -------------------------------------------------------------

# 1. Llamar a la función bubble_sort para ordenar la lista 'scores' in situ
# Después de esta línea, 'scores' está ordenada: [70, 75, 87, 90, 100, 100]
bubble_sort(scores)

# 2. Asignar la lista 'scores' ya ordenada a la nueva variable 'sorted_scores'
sorted_scores = scores 

print("Lista original 'scores' después de ordenar (in situ):")
print(scores)
````

#### Paso 2: Define una función para calcular el promedio de la clase
Define una función llamada `average_class_score` para calcular el promedio de la clase. Agrega un sistema de gestión de errores para los casos en que la lista de estudiantes esté vacía.

````python
### YOUR CODE HERE ###
def average_class_score(scores_list):
    # Gestión de errores: Verificar si la lista está vacía para evitar división por cero
    if len(scores_list) == 0:
        return 0
    
    # Cálculo del promedio: suma total dividida por la cantidad de elementos
    average = sum(scores_list) / len(scores_list)
    
    return average
````
