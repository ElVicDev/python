# Conceptos básicos de Python
## Sentencias de decisión: Controlar el flujo del código

En el núcleo de toda decisión hay una pregunta fundamental: ¿Esta condición es verdadera o falsa? En programación, las sentencias de decisión implementan directamente este concepto fundamental de verdadero/falso, actuando como mecanismo de control del flujo del programa. Evalúan meticulosamente una condición, que puede ir desde una simple comparación.
La sentencia de decisión más básica y utilizada es la sentencia **if**.

```python
if order_total >= 50:
  print("Congratulations! You qualify for free shipping.")
else:
  print("Add more items to your cart to reach the free shipping threshold!")
```

## Enriquecimiento de la selección con cláusulas *elif*

En el ámbito de los escenarios del mundo real, las decisiones a menudo implican más de dos posibilidades. Aquí es donde entra en juego la cláusula **elif**, abreviatura de "*else if*" (si no). A diferencia de una simple sentencia if, una estructura **if-elif-else** permite definir múltiples condiciones y sus correspondientes acciones.
*Imagine que está desarrollando un programa para asignar niveles de afiliación en función de los puntos de fidelidad de los clientes. Puede utilizar las cláusulas **elif** para definir los distintos niveles:*

```python
if points >= 1000:
  print("You've achieved Platinum status! Enjoy exclusive benefits.")
elif points >= 500:
  print("You're a Gold member! Thank you for your loyalty.")
elif points >= 100:
  print("Welcome to the Silver tier! Start earning rewards.")
else:
  print("You're a Bronze member. Keep shopping to earn more points!")
```

# Bucles y a las sentencias condicionales
## Bucle *for*: Iterar sobre secuencias

Un bucle **for** itera sobre una secuencia, como una lista, tupla o cadena, ejecutando un bloque de código por cada elemento de la secuencia. Esto resulta especialmente útil cuando se trabaja con colecciones de datos o cuando se conoce de antemano el número de iteraciones.
*Por ejemplo, para imprimir los 10 primeros números naturales, un bucle **for** puede iterar eficazmente sobre los números del 1 al 10, ejecutando la función print() para cada número.*

```python
for i in range(1, 11):
  print(i)
```

## Bucle *while*: Ejecución basada en condiciones

Un bucle **while**, por el contrario, ejecuta un bloque de código mientras se cumpla una condición especificada. Esto resulta útil cuando se desconoce de antemano el número de iteraciones o cuando el bucle debe continuar hasta que se cumpla una condición específica.
*Por ejemplo, considere un programa que solicite a un usuario una entrada hasta que introduzca una respuesta válida. Un bucle **while** puede pedir repetidamente la entrada y comprobar su validez, terminando sólo cuando se proporciona una entrada válida.*

```python
valid_input = False
while not valid_input:
  user_input = int(input("Please enter a number greater than 0: "))
  if user_input > 0:
    valid_input = True
  else:
    print("Invalid input. Please try again.")
```

## Sentencia *if-else*: Ejecución ramificada

La sentencia **if-else** sigue una estructura sencilla. Si una condición es verdadera, se ejecuta el bloque de código dentro de la sentencia if; en caso contrario, se ejecuta el bloque de código dentro de la sentencia else. Esto permite la ejecución selectiva de código basada en criterios específicos.

*Considere un juego en el que un jugador gana puntos. Una sentencia **if-else** puede mostrar un mensaje de felicitación si la puntuación del jugador supera un determinado umbral, proporcionando comentarios y estímulos basados en su rendimiento.*

```python
player_score = 80

if player_score > 100:
  print("Congratulations! You scored over 100 points.")
else:
  print("Keep trying to beat your high score!")
```
