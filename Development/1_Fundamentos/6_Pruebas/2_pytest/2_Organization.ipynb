## Organización y estructura de pruebas en pytest: Ordenar las pruebas

Un conjunto de pruebas bien estructurado agiliza el proceso de desarrollo. Le ayuda a detectar errores a tiempo, evita regresiones y le da confianza para refactorizar y mejorar el código sin miedo a romper la funcionalidad existente. 

### La importancia de la organización de las pruebas
Imagine que tiene una prueba que falla en un conjunto de pruebas desorganizado. Tiene que vadear a través de un revoltijo de funciones de prueba con nombres crípticos, propósitos poco claros y dependencias enmarañadas. La depuración se convierte en una pesadilla al intentar rastrear el origen del fallo, saltando entre archivos y descifrando código enrevesado.
Por el contrario, un conjunto de pruebas bien organizado es como un sueño placentero. Cada prueba tiene un nombre claro y descriptivo, reside en una ubicación lógica y está perfectamente clasificada. Cuando una prueba falla, puede localizarla rápidamente, comprender su intención e identificar la parte afectada de su código base. Es como tener un bibliotecario que te indica al instante la estantería y el libro exactos que necesitas.
Esta claridad y estructura no sólo le ahorran un tiempo precioso, sino que también reducen la frustración y hacen que el proceso de depuración sea mucho más eficaz. Puede centrarse en solucionar el problema real en lugar de perderse en un laberinto de pruebas desorganizadas.

### Principios clave de la organización de pruebas en pytest
**Modularidad**

La modularidad en las pruebas significa organizar sus pruebas en unidades más pequeñas e independientes. Cada prueba se centra en una parte específica de su código, asegurándose de que funciona correctamente por sí sola. Esto hace que sea más fácil de identificar y solucionar problemas, similar a la solución de problemas de una máquina mediante la comprobación de cada componente individualmente. También permite cambiar o actualizar partes del código sin afectar a todo el conjunto de pruebas, como cuando se sustituye una pieza de una máquina sin desmontarla por completo.
Ilustremos el concepto de modularidad en las pruebas con un ejemplo de código Python que utiliza el módulo **pytest**. Se trata de una sencilla clase de calculadora que demuestra cómo escribir pruebas modulares para sus funciones.


```python
import pytest

class Calculator:
    def add(self, x, y):
        return x + y

    def subtract(self, x, y):
        return x - y

    def multiply(self, x, y):
        return x * y

    def divide(self, x, y):
        if y == 0:
            raise ValueError("Division by zero!")
        return x / y

class CalculatorTest:
    @pytest.fixture
    def calculator(self):
        from calculator import Calculator
        return Calculator()

    def test_add(self, calculator):
        result = calculator.add(2, 3)
        assert result == 5

    def test_subtract(self, calculator):
        result = calculator.subtract(5, 2)
        assert result == 3

    def test_multiply(self, calculator):
        result = calculator.multiply(4, 3)
        assert result == 12

    def test_divide(self, calculator):
        result = self.calculator.divide(10, 2)
        assert result == 5

    def test_divide_by_zero(self, calculator):
        with pytest.raises(ValueError):
            calculator.divide(5, 0)
```

Este ejemplo muestra el concepto de modularidad en las pruebas mediante un sencillo ejemplo de calculadora. La clase **Calculator** define las operaciones aritméticas básicas, mientras que la clase **CalculatorTest** emplea métodos de prueba individuales para examinar cada operación de forma aislada. Observe que el método **Calculator** está precedido por **@pytest.fixture**. El fixture asegura una instancia de Calculadora limpia para cada prueba, y las aserciones verifican el comportamiento esperado. Este enfoque modular permite la identificación precisa de los problemas, ya que cada prueba se centra en una sola operación. La modificación de una función, como **add**, sólo requiere ajustes en su prueba correspondiente, **test_add**. Además, la resolución de problemas es más precisa: un fallo en **test_divide_by_zero** localiza el área problemática. A medida que el código base se amplía, las pruebas modulares facilitan las actualizaciones y la refactorización sin alterar todo el conjunto de pruebas. Aunque este ejemplo es simplificado, el principio de modularidad sigue siendo vital para realizar pruebas eficaces y mantener el código en proyectos reales de mayor complejidad.

**Convenciones de nomenclatura**

Los nombres descriptivos de las funciones y archivos de prueba son esenciales para una organización eficaz de las pruebas. Un nombre claro y específico, como **test_calculate_total_price**, indica inmediatamente la finalidad de la prueba. Le ayuda a localizar y comprender rápidamente la prueba sin necesidad de profundizar en el propio código. Por el contrario, un nombre impreciso como **test_1** no proporciona información sobre el propósito de la prueba y obliga a leer el código para averiguar qué se está comprobando. Los nombres descriptivos actúan como etiquetas para sus pruebas, haciendo que su conjunto de pruebas sea más navegable y fácil de entender, especialmente cuando se trata de un gran número de pruebas o cuando varios desarrolladores están trabajando en el mismo proyecto.

**Fixtures**

Los Fixtures desempeñan un papel importante en la organización de las pruebas, ya que proporcionan un entorno de pruebas estandarizado. Permiten definir código de configuración y desmontaje que puede compartirse en varias pruebas, lo que garantiza la coherencia y reduce la redundancia. Al utilizar fixtures, se evita repetir el mismo código de configuración y desmontaje en cada prueba, lo que hace que el conjunto de pruebas sea más conciso y fácil de mantener. También ayuda a evitar interacciones inesperadas entre las pruebas, lo que conduce a resultados de las pruebas más fiables y predecibles.

**Marcadores**

Los marcadores son una potente herramienta de organización que permite clasificar y filtrar las pruebas en función de sus características. Los marcadores integrados, como **@pytest.mark.slow**, ayudan a identificar las pruebas lentas o dependientes de la red, mientras que los marcadores personalizados permiten agrupar las pruebas por características o funcionalidades. Mediante la aplicación estratégica de marcadores, puede ejecutar selectivamente subconjuntos específicos de pruebas, optimizando sus esfuerzos de comprobación y ahorrando tiempo y recursos valiosos.

**Parametrización**

La parametrización en **pytest** le permite probar su código ejecutando la misma función de prueba varias veces con diferentes valores de entrada, asegurándose de que funciona correctamente en diferentes condiciones. Este enfoque sistemático le ayuda a descubrir posibles errores o comportamientos inesperados que podrían surgir cuando su código se encuentra con una variedad de escenarios y le ayuda a identificar y abordar posibles problemas antes de que afecten a los usuarios.

**Ejemplos prácticos y escenarios**

Exploremos cómo la organización y la estructura de las pruebas pueden funcionar en un escenario del mundo real. Imagínese que está construyendo una plataforma de comercio electrónico, repleta de actividad a medida que los clientes examinan los productos, los añaden a sus carritos y completan las compras. Exploremos cómo podemos aplicar los principios de pytest para crear un conjunto de pruebas bien estructurado y mantenible.

**Funciones del producto**

Dentro de su aplicación de comercio electrónico, un conjunto básico de funcionalidades gira en torno a las interacciones con los productos. Querrá asegurarse de que los clientes puedan añadir artículos a su cesta sin problemas. Para comprobar esto, puede tener una función de prueba llamada **test_add_to_cart_successful** que verifica si un producto se añade correctamente al carrito cuando un usuario hace clic en el botón "Añadir al carrito". Otro aspecto crítico es calcular el precio total, incluyendo impuestos y gastos de envío. Esto podría comprobarse con una función como **test_calculate_total_price_accuracy**, que comprueba si el total calculado coincide con el valor esperado en función de los precios de los productos y los descuentos aplicables.
Hablando de descuentos, es probable que tenga ofertas promocionales o programas de fidelización que ofrezcan descuentos a los clientes. En este caso, una función de prueba denominada **test_apply_discount_correctly** podría verificar si el descuento se aplica correctamente, dando como resultado el precio final correcto. Por último, el proceso de pago es crucial. Podría tener una prueba como **test_checkout_successful_payment** para asegurarse de que la integración de la pasarela de pago funciona sin problemas, y el pedido se realiza correctamente.

Este código Python implementa funcionalidades básicas de comercio electrónico como añadir productos a un carrito, calcular totales con impuestos y descuentos, y simular un proceso de pago, junto con pruebas unitarias para asegurar su corrección:

```python
import pytest

class Product:
    def __init__(self, name, price, quantity=1):
        self.name = name
        self.price = price
        self.quantity = quantity

class Cart:
    def __init__(self):
        self.items = []

    def add_to_cart(self, product):
        self.items.append(product)

    def calculate_total_price(self, tax_rate=0.0, shipping_fee=0.0):
        subtotal = sum(item.price * item.quantity for item in self.items)
        tax = subtotal * tax_rate
        total = subtotal + tax + shipping_fee
        return total

    def apply_discount(self, discount_percentage):
        for item in self.items:
            item.price *= (1 - discount_percentage / 100)

    def checkout(self, payment_gateway):
        total_price = self.calculate_total_price()
        # In a real application, you'd interact with the payment_gateway here
        if payment_gateway.process_payment(total_price):
            # Place the order (logic not shown here)
            return True
        else:
            return False

class TestProductFunctions:
    def test_add_to_cart_successful(self):
        cart = Cart()
        product = Product("Widget", 10.0)
        cart.add_to_cart(product)
        assert len(cart.items) == 1
        assert cart.items[0] == product

    def test_calculate_total_price_accuracy(self):
        cart = Cart()
        cart.add_to_cart(Product("Widget A", 15.0))
        cart.add_to_cart(Product("Widget B", 20.0, 2))  # 2 of these
        total = cart.calculate_total_price(tax_rate=0.07, shipping_fee=5.0)
        expected_total = (15.0 + 20.0 * 2) * 1.07 + 5.0
        assert total == expected_total

    def test_apply_discount_correctly(self):
        cart = Cart()
        cart.add_to_cart(Product("Discounted Item", 50.0))
        cart.apply_discount(20)  # 20% discount
        assert cart.items[0].price == 40

    def test_checkout_successful_payment(self, mocker):
        # Simular una pasarela de pago (sería una integración real en una aplicación de producción)
        mock_payment_gateway = mocker.Mock()
        mock_payment_gateway.proces_payment.return_value = True

        cart = Cart()
        cart.add_to_cart(Product("Something", 100.0))
        result = cart.checkout(mock_payment_gateway)
        assert result is True # Checkout should be successful
```

Este es un ejemplo simplificado. Una aplicación real de comercio electrónico tendría variaciones de productos más complejas, gestión de inventario, seguimiento de pedidos y otras funcionalidades.

**Organización y estructura**

Ahora imagina todas estas funciones de prueba agrupadas en un único y enorme archivo. Rápidamente se convertiría en algo difícil de gestionar. Aquí es donde entran en juego la organización y la estructura. Agrupando las pruebas relacionadas en módulos independientes, como **test_product_functions.py** y **test_user_interactions.py**, se crea un conjunto de pruebas más fácil de navegar y mantener. Los nombres descriptivos de cada función de prueba proporcionan claridad sobre su propósito, lo que facilita que usted y su equipo entiendan el conjunto de pruebas de un vistazo.

Además, las fixture pueden aprovecharse para agilizar el proceso de pruebas. Por ejemplo, puede crear un fixture llamado **empty_cart** que proporcione un carrito de la compra vacío para las pruebas que impliquen añadir productos. Otro fixture, **test_product**, podría crear un producto de muestra con atributos predefinidos como precio y cantidad.

Un conjunto de pruebas bien organizado y estructurado actúa como una red de seguridad para su aplicación de comercio electrónico.
