## Conjunto de herramientas de depuración
### Sentencias de impresión: El narrador de tu código

Las sentencias `print()` son perfectas para realizar pequeñas correcciones rápidas cuando sólo necesitas comprobar una variable o ver hacia dónde se dirige tu código. Proporcionan información inmediata y son ideales para scripts sencillos o problemas aislados. Colocando estratégicamente las sentencias print dentro de tu código, puedes rastrear el flujo de ejecución, inspeccionar los valores de las variables en diferentes etapas y señalar dónde surgen las discrepancias. Digamos que tiene una función para calcular la media de una lista de números:

```python
def calculate_average(numbers):
    total = sum(numbers)
    count = len(numbers)
    return total / count
```

Si el resultado no parece correcto, inserte instrucciones de impresión:

```python
def calculate_average(numbers):
    print("Input numbers:", numbers)
    total = sum(numbers)
    print("Total:", total)
    count = len(numbers)
    print("Count:", count)
    average = total / count
    print("Average:", average)
    return average
```

### Registro: Una crónica detallada de los acontecimientos
Los registros son ideales para hacer un seguimiento de los problemas en programas grandes o de larga duración. Los registros permiten guardar información y revisarla más tarde, lo que resulta útil para identificar patrones o problemas recurrentes. Las bibliotecas de registro, como el módulo de registro integrado de Python, llevan el concepto de las sentencias de impresión al siguiente nivel. Proporcionan una forma estructurada y configurable de registrar eventos, errores y advertencias a lo largo de la ejecución del código.
En lugar de imprimir mensajes directamente en la consola, el registro le permite escribir mensajes en un archivo o en una base de datos. Puede clasificar los mensajes de registro en función de su gravedad (DEBUG, INFO, WARNING, ERROR, CRITICAL) y filtrar los mensajes que se registran en función de sus necesidades de depuración. He aquí un ejemplo:

```python
import logging

logging.basicConfig(filename='app.log', level=logging.DEBUG)
logging.debug('This is a debug message')
logging.info('This is an informational   message')
logging.warning('This is a warning message')
logging.error('This is an error message')
```

### Aserciones: Guardianes de las expectativas
Las aserciones actúan como redes de seguridad en el código. Son sentencias que expresan condiciones que se espera que se cumplan en momentos concretos. Si una aserción falla, es decir, si la condición no se cumple, el programa lanza una excepción, alertando inmediatamente del problema.

Por ejemplo, en una función que calcula el área de un rectángulo, puede afirmar que la longitud y la anchura son positivas:

```python
def calculate_area(length, width):
    assert length > 0, "Length must be positive"
    assert width > 0, "Width must be positive"
    return length * width
```

### Depuradores: La máquina del tiempo de su código
Los depuradores son como máquinas del tiempo para tu código. Son ideales para el código más complejo. Permiten detener la ejecución, inspeccionar el estado actual de las variables y recorrer el código línea a línea, observando cómo cambian los valores.
Los depuradores proporcionan una visibilidad sin precedentes del comportamiento del código en tiempo de ejecución. Son indispensables para:
* Localizar la línea exacta de código en la que se produce un error.
* Entender la lógica compleja trazando el flujo de datos a través de algoritmos intrincados.
* Probar posibles soluciones en un entorno controlado.
