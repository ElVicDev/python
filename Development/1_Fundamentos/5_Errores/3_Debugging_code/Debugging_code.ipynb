##### **Consejos de depuración**

* **Entendiendo el error:** El `TypeError` indica que estás intentando usar el operador de subíndice [] (usado para acceder a elementos dentro de secuencias como listas o cadenas) en un entero, lo cual no está permitido.
* Verifica los tipos de datos: Usa la función `type()` para confirmar el tipo de dato de una variable si no estás seguro.
* Piensa lógicamente: Los enteros representan valores numéricos individuales, por lo que intentar acceder a un "elemento" dentro de ellos no tiene sentido.
* Revisa tu código: Examina cuidadosamente la línea que causa el error y considera qué intentas lograr. ¿Estás usando el enfoque correcto para el tipo de datos con el que trabajas?

```python
# Error deliberado: intentar acceder a un elemento dentro de un entero
first_insect = all_insects_data[0]
# first_digit_of_legs = first_insect['legs'][0]

# Enfoque correcto (comentado para referencia)
first_digit_of_legs = str(first_insect['legs'])[0]

print(f"The first digit of the number of legs the {first_insect['name']} has is: {first_digit_of_legs}")
```

##### **¿Por qué funciona el enfoque correcto?**

El enfoque corregido funciona porque aborda el problema principal del TypeError. A continuación, se detalla:

* Indexación: Una vez que tenemos una representación en cadena del número de patas, podemos usar [0] para acceder a su primer carácter, que corresponde al primer dígito del número original.

En esencia, aprovechamos la naturaleza secuencial de las cadenas para extraer la información deseada de un valor numérico.

Esta técnica es útil cuando se necesita manipular o analizar dígitos individuales dentro de un número. Recuerde que comprender los tipos de datos y sus capacidades es clave para una depuración y resolución de problemas efectivas en Python.

#### **El TypeError - Cadenas**

En esta celda, explorarás otro tipo de `TypeError` al trabajar con cadenas.

***Primero, ejecuta la celda a continuación y revisa el error.***

##### **Consejos de depuración**

* **Entendiendo el error:** El `TypeError` indica que estás intentando combinar (concatenar) una cadena y un entero directamente usando el operador +, lo cual no está permitido en Python.
* Conversión de tipos: Convierte el entero en una cadena usando str() antes de la concatenación.
* f-strings: Considera usar f-strings (literales de cadena formateadas) para una forma más concisa y legible de incrustar variables dentro de cadenas.

##### **¿Sigues atascado?**
* Si sigues atascado, descomenta el código debajo de la siguiente línea para resolver el error: `# Enfoque correcto (comentado como referencia)`

```python
# Error deliberado: intentar concatenar una cadena y un entero directamente
first_insect = all_insects_data[0]
# sentence = "The " + first_insect['name'] + " has " + first_insect['legs'] + " legs."

# Enfoque correcto (comentado para referencia)
sentence = f"The {first_insect['name']} has {first_insect['legs']} legs."

print(sentence)
```

##### **¿Por qué funciona el enfoque correcto?**

El enfoque corregido, `sentence = f"The {first_insect['name']} has {first_insect['legs']} legs."`, funciona a la perfección porque aprovecha la potencia de las cadenas f (literales de cadena formateadas) en Python.

* Cadenas f para un formato elegante de cadenas: Las cadenas f proporcionan una forma concisa y legible de incrustar variables directamente en cadenas. Al encerrar las variables entre llaves {}, sus valores se convierten automáticamente en cadenas y se insertan en la cadena final.

* Conversión de tipo implícita: En nuestro ejemplo, el valor entero `first_insect['legs']` se convierte implícitamente en una cadena al incluirse en la cadena f. Esto elimina la necesidad de una conversión de tipo explícita mediante `str()`, lo que hace que el código sea más limpio y menos propenso a errores.

#### Accediendo a Elementos de una Lista - El Error de Índice

A continuación, intentará acceder a la información sobre un insecto en una posición específica (índice) dentro de su lista de diccionarios de insectos `all_insects_data`. Intentará acceder deliberadamente a un índice fuera de los límites.

***Primero, ejecute la celda a continuación y revise el error.***

##### **Consejos de depuración**
* **Entendiendo el Error:** El `IndexError` indica que está intentando acceder a un elemento en un índice que no existe en la lista.
* Verificar la Longitud de la Lista: Use la función `len()` para obtener el número de elementos en una lista. Recuerde que los índices de lista comienzan en 0, por lo que el último índice válido es `len(list) - 1`.
* Usar sentencias condicionales: Antes de acceder a un elemento en un índice específico, verifique si el índice está dentro del rango válido mediante una sentencia if.

##### **¿Sigue bloqueado?**
* Si sigue bloqueado, descomente el código debajo de la siguiente línea para resolver el error: `# Enfoque correcto (comentado como referencia)`

```python
# Error deliberado: acceder a un índice que está fuera de los límites
# tenth_insect = all_insects_data[10]

# Enfoque correcto (comentado para referencia)
if len(all_insects_data) > 9:  # Check if the 10th insect exists
    tenth_insect = all_insects_data[9]  # Access the 10th insect (index 9)
else:
    print("There are not enough insects in the collection.")

print(tenth_insect)
```

##### **¿Por qué funciona el enfoque correcto?**

El enfoque corregido, que incluye una condición if para comprobar la longitud de la lista antes de acceder a un elemento, evita el IndexError al garantizar que solo se acceda a los elementos dentro del rango válido de la lista.

* `len()` para la longitud de la lista: La función `len(all_insects_data)` devuelve el número total de elementos (diccionarios de insectos) en la lista.

* Indexación basada en cero: Las listas de Python utilizan la indexación basada en cero, lo que significa que el primer elemento está en el índice 0, el segundo en el índice 1, y así sucesivamente. El último índice válido es `len(list) - 1`.

* Acceso condicional: La condición if `len(all_insects_data) > 9` comprueba si hay al menos 10 elementos en la lista. De ser así, accede de forma segura al décimo elemento (en el índice 9). De lo contrario, imprime un mensaje informativo que indica que el insecto solicitado no existe.

#### Acceso a valores del diccionario - El KeyError

En la siguiente celda, intentará acceder al color de un insecto, aunque esta información no esté presente en sus diccionarios de insectos.

***Primero, ejecute la celda a continuación y revise el error.***

##### **Consejos de depuración**

* **Comprenda el error:** El `KeyError` indica que está intentando acceder a una clave que no existe en el diccionario.
* Compruebe la existencia de la clave:
* Use el operador in: if 'key_name' in dictionary:
* Use el método .get(): value = dictionary.get('key_name', default_value) (devuelve default_value si la clave no existe).
* Gestione correctamente las claves faltantes: Proporcione mensajes informativos o valores predeterminados cuando falten claves, en lugar de dejar que su código se bloquee.

##### **¿Sigues atascado?**
* Si sigues atascado, descomenta el código debajo de la siguiente línea para resolver el error: `# Enfoque correcto (comentado como referencia)`

```python
# Error deliberado: acceso a una clave inexistente
first_insect = all_insects_data[0]
# color = first_insect['color']

# Enfoque correcto (comentado para referencia)
if 'color' in first_insect:
    color = first_insect['color']
else:
    print("Color information is not available for this insect.")

print(f"The color of the {first_insect['name']} is: {color}")
```

##### **¿Por qué funciona el enfoque correcto?**

El enfoque corregido, que utiliza el operador in para comprobar la existencia de la clave, evita el error `KeyError` al garantizar que solo se acceda a la clave `color` si está presente en el diccionario `first_insect`.

* Operador `in` para la pertenencia a la clave: La expresión `color` en `first_insect` comprueba si la clave `color` existe en el diccionario `first_insect`. Devuelve `True` si se encuentra la clave y `False` en caso contrario.

* Acceso condicional: La condición `if` `color` en `first_insect` permite ejecutar el código para acceder al valor `color` solo si la clave está presente. Si falta la clave, se imprime un mensaje informativo en lugar de generar un error.

#### Añadiendo nueva información - El error de atributo

A continuación, intentará añadir el color "marrón" a su diccionario `first_insect`. Intentará usar por error el método `.append()`, diseñado para listas, no para diccionarios.

***Primero, ejecute la celda a continuación y revise el error.***

##### **Consejos de depuración**

* **Comprenda el error:** El error `AttributeError` indica que está intentando usar un método (`.append()`) que no existe para el tipo de dato del diccionario.
* Conozca sus estructuras de datos:
* Los diccionarios utilizan pares clave-valor. Para añadir un nuevo par clave-valor, simplemente asigne un valor a una nueva clave: `dictionary['new_key'] = new_value`
* Las listas son conjuntos ordenados de elementos. Para agregar un elemento al final de una lista, use el método `.append()`: `list.append(new_item)`
* Revise la documentación: Si no está seguro de los métodos disponibles para un tipo de dato en particular, consulte la documentación oficial de Python.

```python
# Error deliberado: intentar usar .append() en un diccionario
first_insect = all_insects_data[0]
# first_insect.append('color', 'brown')

# Enfoque correcto (comentado para referencia)
first_insect['color'] = 'brown'

print(first_insect)
```

##### **¿Por qué funciona el enfoque correcto?**

El enfoque corregido, `first_insect['color'] = 'brown'`, añade correctamente la clave 'color' con el valor 'brown' al diccionario `first_insect` porque utiliza la sintaxis correcta para asignar valores a las claves del diccionario.

* Asignación clave-valor: En los diccionarios de Python, se puede añadir un nuevo par clave-valor o actualizar uno existente mediante el operador de asignación `=`. La sintaxis es `dictionary[key] = value`.

* Naturaleza dinámica de los diccionarios: Los diccionarios son mutables, lo que significa que se pueden modificar añadiendo, actualizando o eliminando pares clave-valor una vez creados.

#### Manejo de la división por cero - El error de división por cero

Por último, calculará la razón de patas a alas para cada insecto. Deberá tener cuidado con los insectos sin alas para evitar errores de división por cero.

***Primero, ejecute la celda a continuación y revise el error.***

##### **Consejos de depuración**
* **Comprenda el error:** El error de división por cero ocurre al intentar dividir un número por cero.
* Compruebe si es cero antes de dividir: Use una sentencia `if` para asegurarse de que el divisor no sea cero antes de realizar la división.
* Manejo correcto del caso cero: Proporcione un mensaje apropiado o un cálculo alternativo cuando se encuentre una división por cero.

```python
# Error deliberado: división potencial por cero
for insect in all_insects_data:
    # leg_to_wing_ratio = insect['legs'] / insect['wings']

    # Enfoque correcto (comentado para referencia)
    if insect['wings'] != 0:
        leg_to_wing_ratio = insect['legs'] / insect['wings']
    else:
        leg_to_wing_ratio = "N/A (Insect has no wings)"

    print(f"The leg-to-wing ratio for the {insect['name']} is: {leg_to_wing_ratio}")
```

##### **¿Por qué funciona el enfoque correcto?**

El enfoque corregido, que incluye una condición `if` para comprobar si hay cero alas, evita el `ZeroDivisionError` al evitar la división por cero.

* Comprobación de cero: La condición `if insect['wings'] != 0` comprueba si el número de alas es distinto de cero.

* Cálculo condicional: Si el número de alas es distinto de cero, se calcula e imprime la relación pata-ala.

* Gestión de alas cero: Si el número de alas es cero, se imprime el mensaje `"N/A (El insecto no tiene alas)"` en lugar de realizar la división, lo que soluciona el problema cuando el cálculo no es posible.
