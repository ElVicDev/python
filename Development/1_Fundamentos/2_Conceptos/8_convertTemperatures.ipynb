### Escenario: Controlando datos de temperatura con Python
Eres un analista de datos que trabaja en un proyecto meteorológico. Has recopilado una serie de lecturas de temperatura en grados Celsius,
pero tus herramientas de análisis e informes requieren temperaturas en grados Fahrenheit.
¡Es hora de poner en práctica tus conocimientos de Python!
Crearás un programa que convierte eficientemente una lista de temperaturas en grados Celsius a sus equivalentes en grados Fahrenheit,
demostrando el poder de los bucles y las variables para gestionar las transformaciones de datos del mundo real.

### Resumen de la actividad:
- Crear e inicializar listas para almacenar datos de temperatura.
- Usar un bucle para iterar sobre las temperaturas en grados Celsius.
- Aplicar la fórmula de conversión de grados Celsius a grados Fahrenheit. - Almacene las temperaturas Fahrenheit convertidas en una nueva lista.


```python
# Paso 1: Almacenar datos de temperatura
celsius_temperatures = [0, 10, 25, 32, 100]  # Lista de temperaturas en Celsius
fahrenheit_temperatures = []  # Lista vacía para temperaturas en Fahrenheit

# Impresión del estado inicial de las listas
print("Celsius Temperatures:", celsius_temperatures)
print("Fahrenheit Temperatures:", fahrenheit_temperatures)
```


### Paso 2: Convertir temperaturas
Es hora de convertir tus temperaturas Celsius a Fahrenheit:
- Se ha proporcionado un bucle for para iterar sobre cada valor Celsius en la lista celsius_temperatures.
- Dentro del bucle:
- Aplicar la fórmula de conversión de Celsius a Fahrenheit: F = (C * 9/5) + 32, donde F es la temperatura Fahrenheit y C es la temperatura Celsius.
- Se ha proporcionado el código para almacenar cada valor Fahrenheit calculado en la lista fahrenheit_temperatures mediante el método append().
- Imprimir los resultados


```python
# Listas del paso 1
celsius_temperatures = [0, 10, 25, 32, 100]
fahrenheit_temperatures = []

# Convierte cada temperatura Celsius a Fahrenheit
for celsius in celsius_temperatures:  #Iniciar el bucle
    fahrenheit = (celsius * 9/5) + 32  # Formula de conversión de Celsius a Fahrenheit
    fahrenheit_temperatures.append(fahrenheit)  # Añadir a la lista

# Imprima los resultados (incluida la salida del paso 1)
print("Celsius Temperatures:", celsius_temperatures)
print("Fahrenheit Temperatures:", fahrenheit_temperatures)
```
