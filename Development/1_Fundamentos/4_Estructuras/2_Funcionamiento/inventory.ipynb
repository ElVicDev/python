### Resumen del proyecto:

* **Listas - Gestión de inventario:** Gestiona y actualiza eficientemente nuestra lista de inventario usando listas de Python. Practicaremos agregando nuevos artículos, eliminando productos agotados y asegurándonos de que nuestro inventario esté siempre actualizado.

Para esta actividad, tendrás acceso a un archivo CSV llamado `inventory.csv` que ya se ha creado y contiene algunos artículos iniciales. Sin embargo, aún no se ha cargado en el cuaderno Jupyter.

En la celda siguiente, agrega `inventory.csv` dentro de `pd.read_csv()` para cargar los datos del inventario desde el archivo CSV. Luego, ejecuta la celda, que hará lo siguiente:
* Extraerá los nombres de los productos del DataFrame y los almacenará en una lista llamada `inventory`.

* Imprimirá la lista del inventario para visualizar los datos.

```python
# Importa una potente herramienta llamada "pandas" que usarás para trabajar y organizar datos fácilmente.
import pandas as pd

# Cargar el inventario desde el archivo CSV
inventory_df = pd.read_csv('inventory.csv')

# Extrae los nombres de los productos del DataFrame y guárdalos en una lista.
inventory = inventory_df['product_name'].tolist()

# Imprime la lista de inventario e inspecciona el resultado.
print(inventory) 
```

Observa que el resultado anterior es una lista de artículos del inventario de Furry Foodie.

Resulta que "Cheesy Chompers" ya no es una opción popular en Furry Foodie y has decidido descontinuarlo. Debes actualizar la lista del inventario para reflejar este cambio.

El método `remove()` busca el artículo especificado en la lista y elimina su primera aparición. Si no se encuentra el artículo, genera una excepción ValueError.

En la celda siguiente:

* Usa el método `remove()` para eliminar la primera aparición de "Cheesy Chompers" de la lista `inventory`.

Luego, ejecuta la celda para imprimir el inventario actualizado y ver los cambios.


```python
# Quitar los "Cheesy Chompers"
inventory.remove("Cheesy Chompers") 

print("\nUpdated inventory:")
print(inventory)
```

¡Excelente trabajo! La lista de inventario actualizada ya no muestra "Cheesy Chompers", que se ha eliminado correctamente.

Sin embargo, acaba de llegar un nuevo envío de deliciosas golosinas para mascotas.

La variable `new_items_str` en la celda siguiente contiene una cadena separada por comas que representa estos nuevos artículos. El código proporcionado convierte `new_items_str` en una lista de artículos individuales mediante el método `split()`.

El método `split()` divide una cadena en una lista de subcadenas según un delimitador especificado (en este caso, una coma seguida de un espacio).

En la celda siguiente: * Utilice el método `extend()` para agregar todos los artículos nuevos de la lista `new_items` al final de la lista de inventario existente, añadiendo `new_items` al método `extend()`.

Finalmente, ejecute la celda para imprimir el inventario actualizado y ver los artículos recién agregados.

```python
print("A new shipment of gourmet goodies has arrived!")

# ¡Nuevo envío de productos! (suministrado como una cuerda)
new_items_str = "Squeaky Sausages, Tuna Tidbits, Crunchy Carrots"

# Convierte la cadena en una lista y almacénala en la variable new_items.
new_items = new_items_str.split(", ") 

# Actualiza la lista de inventario añadiendo "new_items" al método extend().
inventory.extend(new_items)

print("\nUpdated inventory:")
print(inventory) 
```

¡Excelente trabajo añadiendo esos nuevos artículos!

Las secciones anteriores se centraron en el uso de listas y tuplas para la gestión de inventario. Si bien estas estructuras de datos son útiles, presentan limitaciones a la hora de acceder y actualizar de forma eficiente la información específica de los productos.

Los diccionarios ofrecen un enfoque más organizado y flexible, ya que permiten almacenar los nombres de los productos como claves y sus niveles de stock correspondientes como valores. Esta estructura clave-valor facilita búsquedas y modificaciones rápidas, lo que hace que los diccionarios sean ideales para gestionar los datos de inventario.

Ejecute la celda siguiente, que utiliza la biblioteca pandas para cargar datos del archivo CSV y convertirlos en un diccionario. Observe la versatilidad de pandas para manejar diferentes estructuras de datos.

```python
import pandas as pd

# Cargar datos de inventario desde CSV
inventory_df = pd.read_csv('inventory.csv')

# Establezca la columna 'product_name' como índice
inventory_df.set_index('product_name', inplace=True)

# Convierte el DataFrame directamente en un diccionario, seleccionando únicamente la columna 'stock_level'.
inventory_dict = inventory_df['stock_level'].to_dict()

# Imprime el inventario
print("Initial Inventory:")
print(inventory_dict)
```

Observe que el resultado anterior ahora es un diccionario donde los nombres de los productos son las claves y los niveles de stock son los valores correspondientes.

Por ejemplo, "Salmon Snacks" es la clave y "25" es el valor.

Esta estructura permite una búsqueda y gestión eficientes de la información del inventario.

Imagine que ha llegado un nuevo envío de productos para mascotas a The Furry Foodie y necesita actualizar su diccionario `inventory_dict` para reflejar estos cambios.

En la celda siguiente:

* Agregue un nuevo artículo, "Puppy Snacks", al diccionario `inventory_dict` con un nivel de stock de 40.

* También recibimos stock adicional de "Cheesy Chompers". Actualice el nivel de stock de "Cheesy Chompers" en el diccionario `inventory_dict` a 20.

A continuación, ejecute la celda para imprimir el diccionario `inventory_dict` actualizado y vea los cambios.

```python
# Añadir "Puppy Snacks" al inventario
inventory_dict["Puppy Snacks"] = 40

# Actualiza "Cheesy Chompers" en el inventario y asígnale el valor de 20.
inventory_dict["Cheesy Chompers"] = 20

print("\nUpdated Inventory:")
print(inventory_dict)
```

Observa el resultado anterior. ¡Se ha añadido Puppy Snacks con un nivel de stock de 40 a tu diccionario! ¡El valor del stock de Cheesy Chompers también se ha actualizado a 20! ¡Excelente!

Lamentablemente, te enteras de que las "Peanut Butter Biscuits" se han descontinuado en The Furry Foodie.

Debes eliminar este artículo de tu diccionario `inventory_dict` para mantener los registros actualizados.

En la celda siguiente:

* Usa la palabra clave `del` para eliminar la entrada "Peanut Butter Biscuits" del diccionario `inventory_dict`. Añade `del` al principio de la variable `inventory_dict` y escribe "Peanut Butter Biscuits".

Luego, ejecuta la celda para imprimir el `inventory_dict` actualizado y ver los cambios.

```python
# Hemos descontinuado las"Peanut Butter Biscuits". Retírelas del inventario.

# Eliminar elemento
del inventory_dict["Peanut Butter Biscuits"] 

print("\nUpdated Inventory after Discontinuation:")
print(inventory_dict)
```

¡Excelente trabajo! Ahora, ejecuta la celda a continuación, que generará un informe de inventario con formato profesional.

Aquí tienes un desglose de la celda final:

* Imprimir encabezado: Imprime un encabezado claro "Informe de inventario:" para diferenciar visualmente el informe de otros resultados.

* Recorrer el inventario: Utiliza un bucle `for` para recorrer cada artículo del diccionario `inventory_dict`.

* Extraer información de producto y stock: En cada iteración del bucle, se utiliza el método `items()` del diccionario para obtener tanto el `producto` (clave) como su nivel de `stock` correspondiente (valor).

* Imprimir salida con formato: Mediante una cadena de texto, crea una línea de texto con formato para cada producto, mostrando su nombre y nivel de stock actual. Esto genera un informe de inventario claro y organizado.

¡Ejecuta el código y consulta el informe!


```python
# ¡Es hora de generar un informe de inventario con un formato adecuado!

print("\nInventory Report:")
for product, stock in inventory_dict.items(): 
    print(f"Product: {product}, Stock: {stock}") 
```
