# Bucles For y While
En programación, la eficiencia y la automatización son fundamentales. Python, un lenguaje versátil célebre por su legibilidad y flexibilidad, ofrece a los programadores potentes herramientas para la repetición: los bucles **for** y **while**.

**for** y **while** mejoran la eficiencia y contribuyen a un código más limpio y fácil de mantener. La distinción clave entre los bucles **for** y **while** es que los bucles **for** iteran sobre una secuencia conocida, mientras que los bucles **while** iteran hasta que una condición se convierte en falsa.

## Bucles For:
Diseñados para iterar sobre secuencias como listas, tuplas, cadenas y rangos. Un bucle **for** puede utilizarse para iterar sobre un rango de números (a menudo denominado bucle "*for i*"), o sobre una secuencia de valores (denominado bucle "*for each*").

**Un punto común de confusión es que el número final no se procesa realmente.**

*Considere este escenario de conteo donde el objetivo es imprimir los números del 1 al 10:*

```python
for i in range(1,11): # This loop prints 1 through 10, not through 11
    print(i)
```

La función **range()** genera una secuencia de números, y el bucle *for* itera sobre esta secuencia. Este bucle *for* continúa ejecutando la sentencia print e incrementa la variable i automáticamente. Estos bucles son especialmente adecuados para situaciones en las que se conoce el número de iteraciones. 

## Bucles while:
Se ejecutan mientras se cumpla una condición especificada. Suelen utilizarse para ejecutarse hasta que se cumpla una condición.

*Por ejemplo, puede implementar el mismo escenario de recuento que se muestra con un bucle for. Observe que es más complicado:*

```python
number = 1
while number <= 10:
    print(number) 
    number = number + 1
```

Este bucle **while** continúa ejecutando la sentencia print e incrementando la variable number mientras se cumpla la condición number <= 10. Aunque los bucles **while** son particularmente útiles cuando el número de iteraciones no se conoce de antemano, ya que el bucle continúa hasta que se cumple una condición específica, *este ejemplo específico se implementa mejor utilizando un bucle **for**.*

Otro ejemplo es la validación de entradas de usuario. Puede utilizar un bucle while para pedir repetidamente a un usuario que introduzca información hasta que proporcione datos en el formato correcto (por ejemplo, una dirección de correo electrónico válida), garantizando que el programa recibe información utilizable.
Considere un ejemplo más sencillo.

*Los usuarios pueden acceder a un menú. Cuando quieren salir del programa, seleccionan la opción 4.*

```python
option = 0
while option != 4:
    print("1. Perform action 1")
    print("2. Perform action 2")
    print("3. Perform action 3")
    print("4. Quit")
```

Dentro del bucle while, normalmente habría más lógica para realizar diferentes acciones basadas en la elección del usuario. Observe también que a la variable option se le asignó un valor de 0.
Mientras no sea 4 inicialmente, el bucle se ejecutará al menos una vez.

# Listas y bucles 
En Python, una lista es una estructura de datos que almacena una colección ordenada de elementos. Estos elementos pueden ser de cualquier tipo. Las listas se definen encerrando los elementos entre corchetes [], separados por comas.
Por ejemplo, una lista que contenga los nombres de las frutas podría declararse como sigue:

    fruits = ["apple", "banana", "cherry"]
    
A cada elemento de una lista se le asigna un índice numérico, empezando por cero. Este índice representa la posición del elemento dentro de la lista.

Hay muchas funciones de lista que verás en futuras lecturas. Una función de uso común es len, que se puede utilizar para determinar cuántos elementos hay en una lista.
Por ejemplo, *el siguiente código imprimiría el número 3, ya que hay tres elementos en la lista de frutas*:

```python
fruits = ["apple", "banana", "cherry"]
fruit_length = len(fruits)
print(fruit_length)
```

A las listas también se les pueden añadir nuevos valores. Por ejemplo, si quieres añadir la fruta 'date' a la lista, puedes utilizar el método **append**:

    fruits = ["apple", "banana", "cherry"]
    fruits.append("date")

El cuarto valor de la lista será ahora date.

Las listas son muy versátiles y facilitan la organización y manipulación de datos dentro de un programa. Una operación común que se realiza en las listas es el recorrido, que implica acceder a cada elemento de forma secuencial. En este caso, los bucles **for** son particularmente adecuados para esta tarea. Este tipo de bucle for se conoce a menudo como bucle **for each**, porque permite recorrer cada elemento individual, de uno en uno.

```python
students = ["Alice", "Bob", "Charlie"]
for student in students:
    print("Hello,", student)
```

Este bucle saluda a cada alumno, mostrando el poder de la iteración en el procesamiento de listas. En este caso, el bucle **for** itera automáticamente sobre cada elemento de la lista students, asignando el elemento actual a la variable student, y ejecutando a continuación la sentencia **print**. 

## Aprovechar los bucles para la iteración
Los bucles **while** son inestimables para tareas con un número desconocido de repeticiones.
*Imagina que quieres simular una tirada de dados hasta obtener un 6:*

```python
import random
roll = 0
while roll != 6:
    roll = random.randint(1, 6) 
    print("You rolled a", roll)
```

## Contar y procesar datos con bucles
Los bucles se integran perfectamente con el recuento y la manipulación de datos.

*Para calcular la suma de números del 1 al 10:*

```python
total = 0
for number in range(1, 11): 
    total += number
    print("The sum is:", total)
```

La función **range()** genera una secuencia de números, y el bucle **for** itera sobre esta secuencia, acumulando la suma en la variable total.

Los bucles **while** pueden realizar tareas más complejas, como filtrar números pares de una lista:

```python
numbers = [1, 2, 3, 4, 5]
index = 0

while index < len(numbers):
    if numbers[index] % 2 == 0:
        print(numbers[index])
    index += 1
```

## Bucles anidados: Amplificación de la iteración
Los bucles son los caballos de batalla de las tareas repetitivas. Sin embargo, el verdadero poder de los bucles brilla cuando están anidados - un bucle dentro de otro. Este anidamiento abre la puerta a la iteración, permitiéndole utilizar estructuras de datos complejas y realizar operaciones intrincadas con facilidad.
*Imagina la tarea de generar una tabla de multiplicar clásica.*
Se trata de una cuadrícula de números en la que cada celda representa el producto de sus números de fila y columna. Los bucles anidados ofrecen una solución elegante. 

```python
for i in range(1, 11):
    for j in range(1, 11):
        print(i, "*", j, "=", i * j, end="\t") # Print the equation
    print() # Move to the next line after each row
```
